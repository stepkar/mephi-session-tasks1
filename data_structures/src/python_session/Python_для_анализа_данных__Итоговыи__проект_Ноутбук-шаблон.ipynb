1. Начнем с простого - с признака **"Образование и ВУЗ"**. Его текущий формат это: **<Уровень образования год выпуска ВУЗ специальность...>**. Например:
* Высшее образование 2016 Московский авиационный институт (национальный исследовательский университет)...
* Неоконченное высшее образование 2000  Балтийская государственная академия рыбопромыслового флота…
Нас будет интересовать только уровень образования.

Создайте с помощью функции-преобразования новый признак **"Образование"**, который должен иметь 4 категории: "высшее", "неоконченное высшее", "среднее специальное" и "среднее".

Выполните преобразование, ответьте на контрольные вопросы и удалите признак "Образование и ВУЗ".

Совет: обратите внимание на структуру текста в столбце **"Образование и ВУЗ"**. Гарантируется, что текущий уровень образования соискателя всегда находится в первых 2ух слов и начинается с заглавной буквы. Воспользуйтесь этим.

*Совет: проверяйте полученные категории, например, с помощью метода unique()*


```python
def extract_education(text):
    if pd.isna(text):
        return np.nan
    tokens = str(text).split()
    first3 = " ".join(tokens[:3]).lower()
    if first3.startswith("неоконченное высшее"):
        return "неоконченное высшее"
    if first3.startswith("среднее специальное"):
        return "среднее специальное"
    if first3.startswith("высшее"):
        return "высшее"
    if first3.startswith("среднее"):
        return "среднее"
    return np.nan

hh["Образование"] = hh["Образование и ВУЗ"].apply(extract_education)
print((hh["Образование"] == "среднее").sum())
hh.drop(columns=["Образование и ВУЗ"], inplace=True)
```

* Женщина , 21 год , родилась 13 января 2000
Как вы понимаете, нам необходимо выделить каждый параметр в отдельный столбец.

Создайте два новых признака **"Пол"** и **"Возраст"**. При этом важно учесть:
* Признак пола должен иметь 2 уникальных строковых значения: 'М' - мужчина, 'Ж' - женщина. 
* Признак возраста должен быть представлен целыми числами.

Выполните преобразование, ответьте на контрольные вопросы и удалите признак **"Пол, возраст"** из таблицы.

*Совет: обратите внимание на структуру текста в столбце, в части на то, как разделены параметры пола, возраста и даты рождения между собой - символом ' , '.

```python
if 'Пол' not in hh.columns or 'Возраст' not in hh.columns:
    if 'Пол, возраст' in hh.columns:
        def _sex(x):
            p = str(x).split(' , ')[0].strip()
            return 'Ж' if p.startswith('Женщина') else ('М' if p.startswith('Мужчина') else np.nan)
        def _age(x):
            second = str(x).split(' , ')[1].strip()
            return int(second.split()[0])
        hh['Пол'] = hh['Пол, возраст'].apply(_sex)
        hh['Возраст'] = hh['Пол, возраст'].apply(_age).astype(int)
        hh.drop(columns=['Пол, возраст'], inplace=True)
print(round(hh['Пол'].eq('Ж').mean()*100, 2))
print(round(pd.to_numeric(hh['Возраст'], errors='coerce').mean(), 1))
```

3. Следующим этапом преобразуем признак **"Опыт работы"**. Его текущий формат - это: **<Опыт работы: n лет m месяцев, периоды работы в различных компаниях…>**. 

Из столбца нам необходимо выделить общий опыт работы соискателя в месяцах, новый признак назовем "Опыт работы (месяц)"

Для начала обсудим условия решения задачи:
* Во-первых, в данном признаке есть пропуски. Условимся, что если мы встречаем пропуск, оставляем его как есть (функция-преобразование возвращает NaN)
* Во-вторых, в данном признаке есть скрытые пропуски. Для некоторых соискателей в столбце стоит значения "Не указано". Их тоже обозначим как NaN (функция-преобразование возвращает NaN)
* В-третьих, нас не интересует информация, которая описывается после указания опыта работы (периоды работы в различных компаниях)
* В-четвертых, у нас есть проблема: опыт работы может быть представлен только в годах или только месяцах. Например, можно встретить следующие варианты:
    * Опыт работы 3 года 2 месяца…
    * Опыт работы 4 года…
    * Опыт работы 11 месяцев…
    * Учитывайте эту особенность в вашем коде

Учитывайте эту особенность в вашем коде

В результате преобразования у вас должен получиться столбец, содержащий информацию о том, сколько месяцев проработал соискатель.
Выполните преобразование, ответьте на контрольные вопросы и удалите столбец **"Опыт работы"** из таблицы.


```python
import re
import numpy as np

def parse_exp_months(x):
    if pd.isna(x):
        return np.nan
    s = str(x).strip().lower()
    if s.startswith("не указано"):
        return np.nan
    # Строго парсим только начало фразы "опыт работы ..."
    m = re.search(r"^\s*опыт\s+работы\s+(?:(\d+)\s+(?:год|года|лет))?(?:\s+(\d+)\s+месяц(?:ев|а)?)?", s)
    if m:
        years = int(m.group(1)) if m.group(1) else 0
        months = int(m.group(2)) if m.group(2) else 0
        return years * 12 + months
    # Фолбэк: берём числа после ключевой фразы, если структура слегка отличается
    y = re.search(r"опыт\s+работы\s+(\d+)\s+(?:год|года|лет)", s)
    mm = re.search(r"опыт\s+работы.*?(\d+)\s+месяц(?:ев|а)?", s)
    years = int(y.group(1)) if y else 0
    months = int(mm.group(1)) if mm else 0
    return years * 12 + months

hh["Опыт работы (месяц)"] = hh["Опыт работы"].apply(parse_exp_months)
print(int(pd.to_numeric(hh["Опыт работы (месяц)"], errors="coerce").median()))
hh.drop(columns=["Опыт работы"], inplace=True)
```

Выполните преобразования и удалите столбец **"Город, переезд, командировки"** из таблицы.

*Совет: обратите внимание на то, что структура текста может меняться в зависимости от указания ближайшего метро. Учите это, если будете использовать порядок слов в своей программе.*


```python
million_cities = ['Новосибирск', 'Екатеринбург','Нижний Новгород','Казань', 'Челябинск','Омск', 'Самара', 'Ростов-на-Дону', 'Уфа', 'Красноярск', 'Пермь', 'Воронеж','Волгоград']

def base_city(x):
    if pd.isna(x):
        return np.nan
    return str(x).split(' , ')[0].strip()

def city_cat(c):
    if pd.isna(c):
        return np.nan
    if c == 'Москва':
        return 'Москва'
    if c == 'Санкт-Петербург':
        return 'Санкт-Петербург'
    if c in million_cities:
        return 'город-миллионник'
    return 'другие'

def can_move(x):
    if pd.isna(x):
        return False
    s = str(x).lower()
    if 'не готов' in s and 'переезд' in s:
        return False
    if 'хочу переехать' in s:
        return True
    if 'готов' in s and 'переезд' in s:
        return True
    return False

def can_trip(x):
    if pd.isna(x):
        return False
    s = str(x).lower()
    if 'не готов' in s and 'командиров' in s:
        return False
    if 'готов' in s and 'командиров' in s:
        return True
    return False

if ('Город' not in hh.columns) or ('Готовность к переезду' not in hh.columns) or ('Готовность к командировкам' not in hh.columns):
    if 'Город, переезд, командировки' in hh.columns:
        hh['Город'] = hh['Город, переезд, командировки'].apply(base_city).apply(city_cat)
        hh['Готовность к переезду'] = hh['Город, переезд, командировки'].apply(can_move)
        hh['Готовность к командировкам'] = hh['Город, переезд, командировки'].apply(can_trip)
        hh.drop(columns=['Город, переезд, командировки'], inplace=True)

print(int(round(hh['Город'].eq('Санкт-Петербург').mean()*100, 0)))
print(int(round((hh['Готовность к переезду'] & hh['Готовность к командировкам']).mean()*100, 0)))
```

5. Рассмотрим поближе признаки **"Занятость"** и **"График"**. Сейчас признаки представляют собой набор категорий желаемой занятости (полная занятость, частичная занятость, проектная работа, волонтерство, стажировка) и желаемого графика работы (полный день, сменный график, гибкий график, удаленная работа, вахтовый метод).

Такой вариант признаков имеет множество различных комбинаций, а значит множество уникальных значений, что мешает анализу. Нужно это исправить!

Давайте создадим признаки-мигалки для каждой категории: если категория присутствует в списке желаемых соискателем, то в столбце на месте строки рассматриваемого соискателя ставится True, иначе - False.

Такой метод преобразования категориальных признаков называется One Hot Encoding и его схема представлена на рисунке ниже:
<img src=https://raw.githubusercontent.com/AndreyRysistov/DatasetsForPandas/main/ohe.jpg>

```python
emp = hh['Занятость'].fillna('').str.get_dummies(sep=', ').astype(bool)
sch = hh['График'].fillna('').str.get_dummies(sep=', ').astype(bool)

hh = pd.concat([hh, emp, sch], axis=1)

print(int((emp.reindex(columns=['проектная работа','волонтерство'], fill_value=False).all(axis=1)).sum()))
print(int((sch.reindex(columns=['вахтовый метод','гибкий график'], fill_value=False).all(axis=1)).sum()))

hh.drop(columns=['Занятость','График'], inplace=True)
```

```python
import re
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Самодостаточный парсер, чтобы исключить влияние старых определений
_def_re = re.compile(r"^\s*опыт\s+работы\s+(?:(\d+)\s+(?:год|года|лет))?(?:\s+(\d+)\s+месяц(?:ев|а)?)?", re.IGNORECASE)

def parse_exp_months_strict(x):
    if pd.isna(x):
        return np.nan
    s = str(x).strip().lower()
    if s.startswith('не указано'):
        return np.nan
    m = _def_re.search(s)
    if m:
        years = int(m.group(1)) if m.group(1) else 0
        months = int(m.group(2)) if m.group(2) else 0
        return years * 12 + months
    y = re.search(r"опыт\s+работы\s+(\d+)\s+(?:год|года|лет)", s)
    mm = re.search(r"опыт\s+работы.*?(\d+)\s+месяц(?:ев|а)?", s)
    years = int(y.group(1)) if y else 0
    months = int(mm.group(1)) if mm else 0
    return years * 12 + months

exp_series = base['Опыт работы'].apply(parse_exp_months_strict)
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(exp_series.dropna(), kde=False, ax=axes[0])
sns.boxplot(x=exp_series.dropna(), ax=axes[1])
plt.tight_layout()
print('mode:', exp_series.mode(dropna=True).iloc[0])
print('min/max:', int(exp_series.min()), int(exp_series.max()))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Убедимся, что в merged есть столбцы "Возраст" и "Образование"
if 'Возраст' not in merged.columns:
    if 'Возраст' in hh.columns and len(hh) == len(merged):
        merged['Возраст'] = hh['Возраст'].values
    elif 'Пол, возраст' in base.columns and len(base) == len(merged):
        def _age_from_str(val):
            try:
                return int(str(val).split(' , ')[1].split()[0])
            except Exception:
                return np.nan
        merged['Возраст'] = base['Пол, возраст'].apply(_age_from_str).values

if 'Образование' not in merged.columns:
    if 'Образование' in base.columns and len(base) == len(merged):
        merged['Образование'] = base['Образование'].values
    elif 'Образование и ВУЗ' in base.columns and 'extract_education' in globals():
        merged['Образование'] = base['Образование и ВУЗ'].apply(extract_education)

heat = merged.groupby(['Возраст','Образование'])['ЗП (руб)'].median().unstack()
plt.figure(figsize=(10,6))
seaborn_plot = sns.heatmap(heat, cmap='viridis')
plt.tight_layout()
# Оценим, у какого уровня образования медианная ЗП растёт быстрее с возрастом
slopes = {}
for edu in heat.columns:
    s = heat[edu].dropna()
    if len(s) > 2:
        x = s.index.values.astype(float)
        y = s.values
        m, b = np.linalg.lstsq(np.vstack([x, np.ones_like(x)]).T, y, rcond=None)[0]
        slopes[str(edu)] = m
best = max(slopes, key=lambda k: slopes[k]) if slopes else None
print('fastest growth education:', best)
```

4. Мы добрались до ликвидации выбросов. Сначала очистим данные вручную. Удалите резюме, в которых указана заработная плата либо выше 1 млн. рублей, либо ниже 1 тыс. рублей.

```python
import pandas as pd
import numpy as np

rates_path = "/Users/stepankarsakov/IdeaProjects/python_learning/untitled/data_structures/src/python_session/ExchangeRates.csv"
fx = pd.read_csv(rates_path)[['currency','date','proportion','close']]
fx['date'] = pd.to_datetime(fx['date'], format='%d/%m/%y').dt.date

# Гарантируем столбец «ЗП (руб)» в hh
if 'ЗП (руб)' not in hh.columns:
    if {'ЗП','Обновление резюме'}.issubset(hh.columns):
        hh_dates = pd.to_datetime(hh['Обновление резюме'], format='%d.%m.%Y %H:%M').dt.date
        zp = hh['ЗП'].astype(str).str.strip()
        amt = zp.str.extract(r'(^[\d\s]+)')[0].str.replace(' ', '', regex=False).astype('float64')
        cur_raw = zp.str.extract(r'([A-Za-zА-Яа-я\.]+)$')[0].str.replace('.', '', regex=False).str.lower()
        cur_map = {
            'руб': 'RUB', 'rur': 'RUB', 'rub': 'RUB',
            'грн': 'UAH', 'uah': 'UAH',
            'usd': 'USD',
            'eur': 'EUR',
            'белруб': 'BYN', 'byn': 'BYN',
            'kgs': 'KGS',
            'сум': 'UZS', 'uzs': 'UZS',
            'azn': 'AZN',
            'kzt': 'KZT'
        }
        cur_iso = cur_raw.map(cur_map).fillna('RUB')
        tmp = pd.DataFrame({'date': hh_dates, 'currency': cur_iso, '_amt': amt})
        m = tmp.merge(fx, on=['date','currency'], how='left')
        mask_rub = m['currency'].eq('RUB')
        m['close'] = np.where(mask_rub & m['close'].isna(), 1, m['close'])
        m['proportion'] = np.where(mask_rub & m['proportion'].isna(), 1, m['proportion'])
        hh['ЗП (руб)'] = m['_amt'] * m['close'] / m['proportion']
    else:
        # Пересчёт по исходной базе и сопоставление по индексу
        src_path = "/Users/stepankarsakov/IdeaProjects/python_learning/untitled/data_structures/src/python_session/dst-3.0_16_1_hh_database.csv"
        base0 = pd.read_csv(src_path, sep=';', low_memory=False)
        base0_dates = pd.to_datetime(base0['Обновление резюме'], format='%d.%m.%Y %H:%M').dt.date
        zp = base0['ЗП'].astype(str).str.strip()
        amt = zp.str.extract(r'(^[\d\s]+)')[0].str.replace(' ', '', regex=False).astype('float64')
        cur_raw = zp.str.extract(r'([A-Za-zА-Яа-я\.]+)$')[0].str.replace('.', '', regex=False).str.lower()
        cur_map = {
            'руб': 'RUB', 'rur': 'RUB', 'rub': 'RUB',
            'грн': 'UAH', 'uah': 'UAH',
            'usd': 'USD',
            'eur': 'EUR',
            'белруб': 'BYN', 'byn': 'BYN',
            'kgs': 'KGS',
            'сум': 'UZS', 'uzs': 'UZS',
            'azn': 'AZN',
            'kzt': 'KZT'
        }
        cur_iso = cur_raw.map(cur_map).fillna('RUB')
        tmp = pd.DataFrame({'date': base0_dates, 'currency': cur_iso, '_amt': amt})
        m = tmp.merge(fx, on=['date','currency'], how='left')
        mask_rub = m['currency'].eq('RUB')
        m['close'] = np.where(mask_rub & m['close'].isna(), 1, m['close'])
        m['proportion'] = np.where(mask_rub & m['proportion'].isna(), 1, m['proportion'])
        zp_rub_full = m['_amt'] * m['close'] / m['proportion']
        hh['ЗП (руб)'] = zp_rub_full.reindex(hh.index).values

hh['ЗП (руб)'] = pd.to_numeric(hh['ЗП (руб)'], errors='coerce')
_before = len(hh)
hh = hh[(hh['ЗП (руб)'] >= 1000) & (hh['ЗП (руб)'] <= 1_000_000)].copy()
_removed = _before - len(hh)
print('removed:', _removed)
print(int(hh['ЗП (руб)'].min()), int(hh['ЗП (руб)'].max()))
```
